```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns when viewing the dataset
pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")
```

```python
# Define the dataset path

file_path = "/kaggle/input/datasets/kmmohammadsinan/global-retail-sales-and-profitability-analysis/global_superstore.csv"

# Load the dataset
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
# Display dataset dimensions

print(f"Number of rows: {df.shape[0]:,}")
print(f"Number of columns: {df.shape[1]}")
```

```python
# Display all column names

print("Dataset Columns:\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
# Display detailed dataset information

df.info()
```

```python
# Create a summary of dataset columns

data_summary = pd.DataFrame({
    'Column': df.columns,
    'Data Type': df.dtypes.astype(str),
    'Non-Null Count': df.notnull().sum().values,
    'Missing Values': df.isnull().sum().values,
    'Unique Values': df.nunique().values
})

data_summary
```

```python
# Calculate missing values and percentages

missing_values = pd.DataFrame({
    'Column': df.columns,
    'Missing Values': df.isnull().sum().values,
    'Percentage (%)': (
        df.isnull().sum().values / len(df) * 100
    ).round(2)
})

# Sort by highest number of missing values
missing_values = missing_values.sort_values(
    by='Missing Values',
    ascending=False
)

missing_values
```

```python
# Visualize missing values

plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap='viridis'
)

plt.title('Missing Values Overview')
plt.xlabel('Columns')

plt.show()
```

```python
# Count completely duplicated rows

duplicate_count = df.duplicated().sum()

duplicate_percentage = (
    duplicate_count / len(df) * 100
)

print(f"Duplicate Rows: {duplicate_count:,}")
print(f"Duplicate Percentage: {duplicate_percentage:.2f}%")
```

```python
# Display the number of unique values in each column

unique_counts = pd.DataFrame({
    'Column': df.columns,
    'Unique Values': df.nunique().values
}).sort_values(by='Unique Values')

unique_counts
```

```python
# Check values in the record count column

df['记录数'].value_counts()
```

```python
# Select key numerical business variables

numerical_columns = [
    'Sales',
    'Profit',
    'Quantity',
    'Discount',
    'Shipping.Cost'
]

# Display descriptive statistics
df[numerical_columns].describe().T
```

```python
# Check zero and negative sales values

zero_sales = (df['Sales'] == 0).sum()
negative_sales = (df['Sales'] < 0).sum()

print(f"Zero Sales Records: {zero_sales:,}")
print(f"Negative Sales Records: {negative_sales:,}")
```

```python
# Check profit distribution

negative_profit = (df['Profit'] < 0).sum()
zero_profit = (df['Profit'] == 0).sum()
positive_profit = (df['Profit'] > 0).sum()

print(f"Negative Profit Records: {negative_profit:,}")
print(f"Zero Profit Records: {zero_profit:,}")
print(f"Positive Profit Records: {positive_profit:,}")
```

```python
# Check quantity values

zero_quantity = (df['Quantity'] == 0).sum()
negative_quantity = (df['Quantity'] < 0).sum()

print(f"Zero Quantity Records: {zero_quantity:,}")
print(f"Negative Quantity Records: {negative_quantity:,}")

print("\nQuantity Range:")
print(f"Minimum Quantity: {df['Quantity'].min()}")
print(f"Maximum Quantity: {df['Quantity'].max()}")
```

```python
# Check discount range

print(f"Minimum Discount: {df['Discount'].min()}")
print(f"Maximum Discount: {df['Discount'].max()}")

print("\nUnique Discount Values:")
print(sorted(df['Discount'].unique()))
```

```python
# Check shipping cost values

zero_shipping_cost = (df['Shipping.Cost'] == 0).sum()
negative_shipping_cost = (df['Shipping.Cost'] < 0).sum()

print(f"Zero Shipping Cost Records: {zero_shipping_cost:,}")
print(f"Negative Shipping Cost Records: {negative_shipping_cost:,}")

print("\nShipping Cost Range:")
print(f"Minimum: {df['Shipping.Cost'].min():.2f}")
print(f"Maximum: {df['Shipping.Cost'].max():.2f}")
```

```python
negative_profit_percentage = (
    (df['Profit'] < 0).sum() / len(df) * 100
)

print(f"Negative Profit Percentage: {negative_profit_percentage:.2f}%")
```

```python
# Investigate records with zero sales

zero_sales_records = df[df['Sales'] == 0]

zero_sales_records
```

```python
# Inspect raw date values

print("Order Date Examples:")
display(df['Order.Date'].head())

print("\nShip Date Examples:")
display(df['Ship.Date'].head())
```

```python
# Convert date columns to datetime format

df['Order.Date'] = pd.to_datetime(
    df['Order.Date'],
    errors='coerce'
)

df['Ship.Date'] = pd.to_datetime(
    df['Ship.Date'],
    errors='coerce'
)

print("Date conversion completed successfully.")
```

```python
# Check date data types and missing values after conversion

print("Data Types:")
print(df[['Order.Date', 'Ship.Date']].dtypes)

print("\nMissing Dates After Conversion:")
print(df[['Order.Date', 'Ship.Date']].isnull().sum())
```

```python
# Check dataset date range

print("Order Date Range")
print(f"Earliest Order Date: {df['Order.Date'].min().date()}")
print(f"Latest Order Date: {df['Order.Date'].max().date()}")

print("\nShip Date Range")
print(f"Earliest Ship Date: {df['Ship.Date'].min().date()}")
print(f"Latest Ship Date: {df['Ship.Date'].max().date()}")
```

```python
# Check for invalid chronological relationships

invalid_shipping_dates = df[
    df['Ship.Date'] < df['Order.Date']
]

print(
    f"Records where shipping occurred before ordering: "
    f"{len(invalid_shipping_dates):,}"
)
```

```python
# Calculate shipping duration in days

df['Shipping.Days'] = (
    df['Ship.Date'] - df['Order.Date']
).dt.days

# Summary statistics
df['Shipping.Days'].describe()
```

```python
# Validate shipping duration

negative_shipping_days = (
    df['Shipping.Days'] < 0
).sum()

print(
    f"Negative Shipping Duration Records: "
    f"{negative_shipping_days}"
)
```

```python
# Extract year directly from order date

extracted_year = df['Order.Date'].dt.year

# Compare existing Year column with extracted year

year_mismatches = (
    df['Year'] != extracted_year
).sum()

print(f"Year mismatches: {year_mismatches:,}")
```

```python
# Extract ISO week number from Order.Date

extracted_week = (
    df['Order.Date']
    .dt.isocalendar()
    .week
)

# Compare with existing week number

week_mismatches = (
    df['weeknum'].astype(int) != extracted_week.astype(int)
).sum()

print(f"Week Number mismatches: {week_mismatches:,}")
```

```python
# Create a copy of the dataset

df_clean = df.copy()

print("Working copy created successfully!")
print(f"Shape: {df_clean.shape}")
```

```python
# Remove constant record count column

df_clean.drop(columns=['记录数'], inplace=True)

print("Constant column removed successfully.")
print(f"Updated Shape: {df_clean.shape}")
```

```python
# Remove records with zero sales

rows_before = len(df_clean)

df_clean = df_clean[df_clean['Sales'] > 0].copy()

rows_after = len(df_clean)

print(f"Rows before removal: {rows_before:,}")
print(f"Rows after removal: {rows_after:,}")
print(f"Rows removed: {rows_before - rows_after}")
```

```python
# Rename columns using a consistent naming convention

df_clean.rename(columns={
    'Category': 'category',
    'City': 'city',
    'Country': 'country',
    'Customer.ID': 'customer_id',
    'Customer.Name': 'customer_name',
    'Discount': 'discount',
    'Market': 'market',
    'Order.Date': 'order_date',
    'Order.ID': 'order_id',
    'Order.Priority': 'order_priority',
    'Product.ID': 'product_id',
    'Product.Name': 'product_name',
    'Profit': 'profit',
    'Quantity': 'quantity',
    'Region': 'region',
    'Row.ID': 'row_id',
    'Sales': 'sales',
    'Segment': 'segment',
    'Ship.Date': 'ship_date',
    'Ship.Mode': 'ship_mode',
    'Shipping.Cost': 'shipping_cost',
    'State': 'state',
    'Sub.Category': 'sub_category',
    'Year': 'year',
    'Market2': 'market_group',
    'weeknum': 'week_number',
    'Shipping.Days': 'shipping_days'
}, inplace=True)

print("Column names standardized successfully!")
```

```python
print(df_clean.columns.tolist())
```

```python
# Regenerate year

df_clean['year'] = df_clean['order_date'].dt.year

# Generate month number
df_clean['month'] = df_clean['order_date'].dt.month

# Generate month name
df_clean['month_name'] = df_clean['order_date'].dt.month_name()

# Generate quarter
df_clean['quarter'] = df_clean['order_date'].dt.quarter

# Generate ISO week number
df_clean['week_number'] = (
    df_clean['order_date']
    .dt.isocalendar()
    .week
    .astype(int)
)

# Generate day of week
df_clean['day_of_week'] = df_clean['order_date'].dt.day_name()

print("Time-based features generated successfully!")
```

```python
# Calculate profit margin percentage

df_clean['profit_margin'] = (
    df_clean['profit'] / df_clean['sales']
) * 100

df_clean['profit_margin'] = df_clean['profit_margin'].round(2)

df_clean[['sales', 'profit', 'profit_margin']].head()
```

```python
# Convert discount to percentage

df_clean['discount_percentage'] = (
    df_clean['discount'] * 100
).round(2)

df_clean[['discount', 'discount_percentage']].head()
```

```python
# Final dataset overview

print("Final Dataset Shape:")
print(df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df_clean.duplicated().sum())

print("\nZero Sales Records:")
print((df_clean['sales'] == 0).sum())

print("\nNegative Shipping Days:")
print((df_clean['shipping_days'] < 0).sum())
```
